```python
import requests

api_key = "YOUR_API_KEY"

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": "Ahmedabad",
    "appid": api_key,
    "units": "metric"
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    temperature = data["main"]["temp"]
    description = data["weather"][0]["description"]

    print("City: Ahmedabad")
    print("Temperature:", temperature, "°C")
    print("Weather:", description)
else:
    print("Error:", response.status_code)
```

##### TASK - 2

```python
import requests
from datetime import datetime

url = "https://api.coingecko.com/api/v3/simple/price"

params = {
    "ids": "bitcoin,ethereum",
    "vs_currencies": "usd"
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    bitcoin = data["bitcoin"]["usd"]
    ethereum = data["ethereum"]["usd"]

    now = datetime.now()

    print("Date and Time:", now)
    print("Bitcoin Price: $", bitcoin)
    print("Ethereum Price: $", ethereum)
else:
    print("Error:", response.status_code)
```

```python
import requests

api_key = "DEMO_KEY"

url = "https://api.nasa.gov/planetary/apod"

params = {
    "api_key": api_key
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    print("Title:", data["title"])
    print("\nExplanation:")
    print(data["explanation"])

    image_url = data["url"]

    image = requests.get(image_url)

    with open("nasa_apod.jpg", "wb") as file:
        file.write(image.content)

    print("\nImage saved as nasa_apod.jpg")
else:
    print("Error:", response.status_code)
```

```python
import requests

url = "https://disease.sh/v3/covid-19/countries/India"

response = requests.get(url)

if response.status_code == 200:
    data = response.json()

    print("+----------------+----------------+")
    print("| Information    | Number         |")
    print("+----------------+----------------+")
    print(f"| Total Cases    | {data['cases']:<14} |")
    print(f"| Recovered      | {data['recovered']:<14} |")
    print(f"| Deaths         | {data['deaths']:<14} |")
    print("+----------------+----------------+")
else:
    print("Error:", response.status_code)
```

```python
import requests

weather_api_key = "YOUR_API_KEY"

weather_url = "https://api.openweathermap.org/data/2.5/weather"

weather_params = {
    "q": "Mumbai",
    "appid": weather_api_key,
    "units": "metric"
}

bitcoin_url = "https://api.coingecko.com/api/v3/simple/price"

bitcoin_params = {
    "ids": "bitcoin",
    "vs_currencies": "usd"
}

try:
    weather_response = requests.get(weather_url, params=weather_params)

    if weather_response.status_code == 200:
        weather_data = weather_response.json()
        temperature = weather_data["main"]["temp"]
        print("Mumbai Temperature:", temperature, "°C")
    else:
        print("Weather API failed:", weather_response.status_code)

except requests.exceptions.RequestException:
    print("Unable to fetch weather data")

try:
    bitcoin_response = requests.get(bitcoin_url, params=bitcoin_params)

    if bitcoin_response.status_code == 200:
        bitcoin_data = bitcoin_response.json()
        bitcoin_price = bitcoin_data["bitcoin"]["usd"]
        print("Bitcoin Price: $", bitcoin_price)
    else:
        print("Bitcoin API failed:", bitcoin_response.status_code)

except requests.exceptions.RequestException:
    print("Unable to fetch Bitcoin data")
```
