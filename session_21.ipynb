```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price - (self.price * 10 / 100)


p1 = Product("Shoes", 2000)

print("Product:", p1.name)
print("Original Price:", p1.price)
print("Discounted Price:", p1.get_discounted_price())
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price - (self.price * 10 / 100)


class Electronics(Product):

    def get_discounted_price(self):
        return self.price - (self.price * 20 / 100)


p1 = Product("Shoes", 2000)
e1 = Electronics("Laptop", 50000)

print("Product Discounted Price:", p1.get_discounted_price())
print("Electronics Discounted Price:", e1.get_discounted_price())
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price - (self.price * 10 / 100)


class Electronics(Product):

    def get_discounted_price(self):
        return self.price - (self.price * 20 / 100)


def show_final_price(item):
    print("Name:", item.name)
    print("Final Price:", item.get_discounted_price())
    print()


p1 = Product("Shoes", 2000)
e1 = Electronics("Laptop", 50000)

show_final_price(p1)
show_final_price(e1)
```

```python
class Ticket:
    def __init__(self, movie_name, price):
        self.movie_name = movie_name
        self.price = price

    def get_final_price(self):
        return self.price


class PremiumTicket(Ticket):

    def get_final_price(self):
        return super().get_final_price() + 50


t1 = Ticket("Avengers", 200)
t2 = PremiumTicket("Avengers", 300)

print("Normal Ticket Price:", t1.get_final_price())
print("Premium Ticket Price:", t2.get_final_price())
```
