````python
# 1.Write a Python function called get_song_duration_per_minute that divides the total duration of a Spotify playlist (in minutes) by the number of songs, and handles the case where the number of songs is zero using try, except, and finally blocks.
def get_song_duration_per_minute(total_duration, num_songs):
    try:
        duration_per_song = total_duration / num_songs
        return duration_per_song
    except ZeroDivisionError:
        return "Number of songs cannot be zero."
    finally:
        print("Calculation completed.")

listening_time = int(input("Enter total duration of the playlist in minutes: "))
num_songs = int(input("Enter the number of songs in the playlist: "))

result = get_song_duration_per_minute(listening_time, num_songs)
print(f"Average duration per song: {result}")
````

````python
class Ticket:
    def __init__(self, movie_name, price):
        self.movie_name = movie_name
        self.price = price

    def get_final_price(self):
        return self.price


class PremiumTicket(Ticket):

    def get_final_price(self):
        return super().get_final_price() + 50


t1 = Ticket("Avengers", 200)
t2 = PremiumTicket("Avengers", 300)

print("Normal Ticket Price:", t1.get_final_price())
print("Premium Ticket Price:", t2.get_final_price())
````

````python
import asyncio
import httpx

async def fetch_data(client, url):
    response = await client.get(url)
    return response.json()

async def main():
    url1 = "https://jsonplaceholder.typicode.com/users/1"
    url2 = "https://jsonplaceholder.typicode.com/posts/1"

    async with httpx.AsyncClient() as client:
        result1, result2 = await asyncio.gather(
            fetch_data(client, url1),
            fetch_data(client, url2)
        )

    print("API 1 Result:")
    print(result1)

    print("\nAPI 2 Result:")
    print(result2)

await main()
````

````python
import requests

weather_api_key = "YOUR_API_KEY"

weather_url = "https://api.openweathermap.org/data/2.5/weather"

weather_params = {
    "q": "Mumbai",
    "appid": weather_api_key,
    "units": "metric"
}

bitcoin_url = "https://api.coingecko.com/api/v3/simple/price"

bitcoin_params = {
    "ids": "bitcoin",
    "vs_currencies": "usd"
}

try:
    weather_response = requests.get(weather_url, params=weather_params)

    if weather_response.status_code == 200:
        weather_data = weather_response.json()
        temperature = weather_data["main"]["temp"]
        print("Mumbai Temperature:", temperature, "°C")
    else:
        print("Weather API failed:", weather_response.status_code)

except requests.exceptions.RequestException:
    print("Unable to fetch weather data")

try:
    bitcoin_response = requests.get(bitcoin_url, params=bitcoin_params)

    if bitcoin_response.status_code == 200:
        bitcoin_data = bitcoin_response.json()
        bitcoin_price = bitcoin_data["bitcoin"]["usd"]
        print("Bitcoin Price: $", bitcoin_price)
    else:
        print("Bitcoin API failed:", bitcoin_response.status_code)

except requests.exceptions.RequestException:
    print("Unable to fetch Bitcoin data")
````

````python
import requests
import csv

url = "https://api.coingecko.com/api/v3/coins/markets"

params = {
    "vs_currency": "usd",
    "order": "market_cap_desc",
    "per_page": 10,
    "page": 1
}

try:
    response = requests.get(url, params=params, timeout=10)

    if response.status_code == 200:
        data = response.json()

        print("Top 10 Cryptocurrencies")
        print("-" * 80)

        for coin in data:
            print("Name:", coin["name"])
            print("Price: $", coin["current_price"])
            print("24h Change:", coin["price_change_percentage_24h"], "%")
            print("24h High: $", coin["high_24h"])
            print("24h Low: $", coin["low_24h"])
            print("-" * 80)

        with open("crypto_prices.csv", "w", newline="") as file:
            writer = csv.writer(file)

            writer.writerow([
                "Name",
                "Current Price",
                "Price Change %",
                "24h High",
                "24h Low"
            ])

            for coin in data:
                writer.writerow([
                    coin["name"],
                    coin["current_price"],
                    coin["price_change_percentage_24h"],
                    coin["high_24h"],
                    coin["low_24h"]
                ])

        print("Data saved successfully to crypto_prices.csv")

    else:
        print("API request failed. Status code:", response.status_code)

except requests.exceptions.RequestException:
    print("Unable to connect to the CoinGecko API.")

except Exception as e:
    print("An error occurred:", e)
````
