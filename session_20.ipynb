```python
class Playlist:
    def __init__(self):
        self._songs = []

    def add_song(self, song):
        self._songs.append(song)

    def show_playlist(self):
        print("Playlist:", self._songs)


playlist = Playlist()

playlist.add_song("Kesariya")
playlist.add_song("Perfect")
playlist.add_song("Tum Hi Ho")

playlist.show_playlist()
```

```python
class Product:
    def __init__(self, price):
        self._price = price

    def get_price(self):
        return self._price

    def set_price(self, price):
        self._price = price


product = Product(25000)

print("Old Price:", product.get_price())

product.set_price(22000)

print("New Price:", product.get_price())
```

```python
class Movie:
    def __init__(self, rating):
        self._rating = rating

    def get_rating(self):
        return self._rating

    def set_rating(self, rating):
        if rating >= 0 and rating <= 10:
            self._rating = rating
        else:
            print("Error: Rating must be between 0 and 10")


movie = Movie(8.5)

print("Rating:", movie.get_rating())

movie.set_rating(9.2)

print("Updated Rating:", movie.get_rating())

movie.set_rating(15)
```

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class Paytm(PaymentMethod):

    def pay(self, amount):
        print("Paid", amount, "using Paytm")


class PhonePe(PaymentMethod):

    def pay(self, amount):
        print("Paid", amount, "using PhonePe")


p1 = Paytm()
p2 = PhonePe()

p1.pay(500)
p2.pay(1000)
```
